```python
# generate a simple pahntom object containing two states
obj_sz = 128
num_angle = 61
V = shepp_logan(obj_sz)
ind1 = np.where(np.abs(V-0.3)<0.01)
mask1 = np.zeros(V.shape)
mask1[ind1] = 1

ind2 = np.where(np.abs(V-0.3)>=0.01)
mask2 = np.zeros(V.shape)
mask2[ind2] = 1

im1 = V*mask1
im2 = V*mask2

plt.subplot(121)
plt.title('state 1')
plt.imshow(im1, interpolation='none')
plt.subplot(122)
plt.imshow(im2,interpolation='none')
plt.title('state 2')
```

```python
import time
import os
obj_row, obj_col = np.shape(V) 
# loading reference spectra
# ref_file = f"{os.getcwd()}\\Experimental data\\NMC811\\Co-used-spect.txt"
ref_file = 'fe0_fe3_ref.txt'
ref_spectra = np.loadtxt(ref_file)
used_spect = np.abs(ref_spectra[:,1:3])


# generate the list of angle-energy pairs and the codebook 
# that records the relative intensity change of each state based on reference spectra
# 'uniform-random' refers to uniform angular step and random energy selection
# other options used in the paper are 'uniform-interlaced' and 'uniform-segmented'

# interlaced-stochastic
# np.random.shuffle(used_spect)

mse_list = []


angle, codebook, ind = generate_plan(used_spect,[0,np.pi],num_angle,'uniform-random')
angle = np.linspace(0,np.pi,num_angle)
sino = []


for i in range (len(angle)):
    proj_geom = astra.create_proj_geom('parallel',1.0,obj_col,angle[i])
    vol_geom = astra.create_vol_geom(obj_row,obj_col)
    proj_id = astra.create_projector('linear',proj_geom,vol_geom)
    [sigogram_id, sinogram] = astra.create_sino(im1*codebook[i,0]+im2*codebook[i,1],proj_id)
    sino.append(sinogram)
sino = np.squeeze(np.array(sino))
    # sino_noise=np.random.poisson(np.array(sino*1e3/np.max(sino),dtype='int'))/1e3

for i in range(1):
    # use random initialization
    x_init = []
    x_init.append(np.random.rand(obj_row,obj_col))
    x_init.append(np.random.rand(obj_row,obj_col))
#     x_init.append(np.random.rand(obj_row,obj_col))

    # note: computation is not parallerized and can take up to a few minutes 
    start_time= time.time()
    rec,err = multistate_tomo_joint_TV(sino,angle,0,\
                                       [obj_row,obj_col],codebook,\
                                       0.1,1,200,x0=x_init,seq_save=True,method='gmres')
    end_time = time.time()
    rec_mse = mse(im1,rec[-1][0]) + mse(im2,rec[-1][1])
    mse_list.append(rec_mse)
    print(f"mean square error = {rec_mse}")

# np.savetxt('uniform-random_mse_20_trials_fixed_spect_random_init.txt',mse_list)
print(f"computation time = {end_time-start_time}")
plt.figure(figsize=(15,5))
plt.subplot(131)
plt.imshow(sino, interpolation = 'none')
plt.title('sinogram')
plt.subplot(132)
plt.imshow(im1,clim=[0,0.3], interpolation = 'none')
plt.colorbar()
plt.title('State 1 (GT)')
plt.xticks([])
plt.yticks([])
plt.subplot(133)
plt.imshow(im2,clim=[0,1], interpolation = 'none')
plt.colorbar()
plt.title('State 2 (GT)')
plt.xticks([])
plt.yticks([])
plt.tight_layout()
plt.figure(figsize=(15,5))
plt.subplot(131)
plt.plot(np.log10(err))
plt.title('convergence')
plt.subplot(132)
plt.imshow(rec[-1][0],clim=[0,0.3], interpolation = 'none')
plt.colorbar()
plt.title('State 1')
plt.xticks([])
plt.yticks([])
plt.subplot(133)
plt.imshow(rec[-1][1],clim=[0,1], interpolation = 'none')
plt.colorbar()
plt.title('State 2')
plt.xticks([])
plt.yticks([])
plt.tight_layout()



```

```python
from itertools import combinations

# ---------- Utilities ----------

def cosine_sim(a, b):
    """Cosine similarity between 1D arrays a and b (dot of L2-normalized vectors)."""
    a = np.asarray(a); b = np.asarray(b)
    na = np.linalg.norm(a); nb = np.linalg.norm(b)
    if na == 0 or nb == 0:
        return np.nan  # undefined if a or b is all zeros
    return float(np.dot(a, b) / (na * nb))

def _score(x, y, inds):
    """Cosine score for a subset of indices; treat NaN as +inf (bad)."""
    r = cosine_sim(x[inds], y[inds])
    return np.inf if np.isnan(r) else r

# ---------- Greedy backward elimination (robust default) ----------

def min_cosine_subset_backward(x, y, k):
    """
    Start with all indices; iteratively remove the single index whose removal
    yields the smallest cosine similarity. Returns the kept indices (length k).
    """
    x = np.asarray(x); y = np.asarray(y)
    n = len(x)
    if not (2 <= k <= n):
        raise ValueError("k must be between 2 and len(x)")

    keep = list(range(n))
    while len(keep) > k:
        best_score = None
        best_to_remove = None
        for idx in keep:
            cand = [j for j in keep if j != idx]
            s = _score(x, y, cand)
            if (best_score is None) or (s < best_score):
                best_score = s
                best_to_remove = idx
        keep.remove(best_to_remove)
    return keep

# ---------- Greedy forward selection (alternative) ----------

def min_cosine_subset_forward(x, y, k):
    """
    Start empty; add the index that minimizes cosine at each step.
    Useful when backward gets stuck differently. Returns kept indices.
    """
    x = np.asarray(x); y = np.asarray(y)
    n = len(x)
    if not (2 <= k <= n):
        raise ValueError("k must be between 2 and len(x)")

    remaining = set(range(n))
    # Seed with the single index whose self-pair has the most negative sign
    # (equivalently, minimal cosine for length-1 — just sign of x_i*y_i).
    best_first = min(remaining, key=lambda i: _score(x, y, [i]))
    keep = [best_first]
    remaining.remove(best_first)

    while len(keep) < k:
        best_score = None
        best_add = None
        for i in remaining:
            cand = keep + [i]
            s = _score(x, y, cand)
            if (best_score is None) or (s < best_score):
                best_score = s
                best_add = i
        keep.append(best_add)
        remaining.remove(best_add)
    return keep

# ---------- 2-opt local improvement (refinement step) ----------

def two_opt_improve(x, y, keep, max_passes=3):
    """
    Try single swaps between kept and not-kept indices if they reduce cosine.
    Stop when no improvement or max_passes reached.
    """
    x = np.asarray(x); y = np.asarray(y)
    keep = set(keep)
    others = set(range(len(x))) - keep

    improved = True
    passes = 0
    while improved and passes < max_passes:
        improved = False
        passes += 1
        base_score = _score(x, y, sorted(keep))
        for out in list(keep):
            for inn in list(others):
                cand_keep = (keep - {out}) | {inn}
                s = _score(x, y, sorted(cand_keep))
                if s < base_score:
                    keep.remove(out); keep.add(inn)
                    others.remove(inn); others.add(out)
                    base_score = s
                    improved = True
        # loop continues if improved
    return sorted(keep)

# ---------- Brute force (for tiny n; sanity check) ----------

def min_cosine_subset_bruteforce(x, y, k):
    """
    Exact solution by trying all k-combinations. Use only for very small n.
    Returns (best_indices, best_cosine).
    """
    x = np.asarray(x); y = np.asarray(y)
    best_inds, best_s = None, None
    for inds in combinations(range(len(x)), k):
        s = _score(x, y, inds)
        if (best_s is None) or (s < best_s):
            best_s = s
            best_inds = list(inds)
    return best_inds, best_s

# ---------- End-to-end helper ----------

def min_cosine_subset(x, y, k, method="backward+2opt"):
    """
    High-level API. Returns (x_sub, y_sub, indices, cosine_score).
    method: "backward", "forward", "backward+2opt", or "forward+2opt"
    """
    if method.startswith("backward"):
        inds = min_cosine_subset_backward(x, y, k)
    elif method.startswith("forward"):
        inds = min_cosine_subset_forward(x, y, k)
    else:
        raise ValueError("Unknown method")

    if method.endswith("+2opt"):
        inds = two_opt_improve(x, y, inds)

    x = np.asarray(x); y = np.asarray(y)
    s = _score(x, y, inds)
    return x[inds], y[inds], inds, s 
```

```python
# add noise to the sinogram
# For very noisy data, mu1 and mu2 need to be tuned more for the best result

ind1 = np.where(np.abs(V-0.3)<0.01)
mask1 = np.zeros(V.shape)
mask1[ind1] = 1

ind2 = np.where(np.abs(V-0.3)>=0.01)
mask2 = np.zeros(V.shape)
mask2[ind2] = 1

im1 = V*mask1
im2 = V*mask2

obj_row, obj_col = np.shape(V) 
# loading reference spectra
ref_spectra = np.loadtxt('fe0_fe3_ref.txt')
used_spect = np.abs(ref_spectra[:,1:3])

num_angle = 181

# generate the list of angle-energy pairs and the codebook 
# that records the relative intensity change of each state based on reference spectra
# 'uniform-random' refers to uniform angular step and random energy selection
# other options used in the paper are 'uniform-interlaced' and 'uniform-segmented'

angle, codebook, ind = generate_plan(used_spect,[0,np.pi],num_angle,'uniform-random')
angle = np.linspace(0,np.pi,num_angle)
sino = []


for i in range (len(angle)):
    proj_geom = astra.create_proj_geom('parallel',1.0,obj_col,angle[i])
    vol_geom = astra.create_vol_geom(obj_row,obj_col)
    proj_id = astra.create_projector('linear',proj_geom,vol_geom)
    [sigogram_id, sinogram] = astra.create_sino(im1*codebook[i,0]+im2*codebook[i,1],proj_id)
    sino.append(sinogram)
sino = np.squeeze(np.array(sino))
maxVal = sino.max()
sino_noise=np.random.poisson(np.array(sino*2e2/maxVal,dtype='int'))*maxVal/2e2

# use random initialization
x_init = []
x_init.append(np.random.rand(obj_row,obj_col))
x_init.append(np.random.rand(obj_row,obj_col))

# note: computation is not parallerized and can take up to a few minutes 
rec,err = multistate_tomo_joint_TV(sino_noise,angle,0,\
                                   [obj_row,obj_col],codebook,\
                                   15,25,300,x0=x_init,seq_save=True)

plt.figure(figsize=(15,5))
plt.subplot(131)
plt.imshow(sino_noise)
plt.title('sinogram')
plt.subplot(132)
plt.imshow(im1)
plt.colorbar()
plt.title('State 1 (GT)')
plt.xticks([])
plt.yticks([])
plt.subplot(133)
plt.imshow(im2)
plt.colorbar()
plt.title('State 2 (GT)')
plt.xticks([])
plt.yticks([])
plt.tight_layout()
plt.figure(figsize=(15,5))
plt.subplot(131)
plt.plot(np.log10(err))
plt.title('convergence')
plt.subplot(132)
plt.imshow(rec[-1][0])
plt.colorbar()
plt.title('State 1')
plt.xticks([])
plt.yticks([])
plt.subplot(133)
plt.imshow(rec[-1][1])
plt.colorbar()
plt.title('State 2')
plt.xticks([])
plt.yticks([])
plt.tight_layout()
```
