```python
# loading experimental data of the test object
data_path = 'Experimental data\\Test object\\'

# load the measured projection data
proj = tf.imread(f'{data_path}Fe_proj.tif')

# load the experimental plan that contains the information about angle-enery pair used in the measurement
plan = np.loadtxt(f'{data_path}plan_52.txt')

# load the angle list
angle = np.deg2rad(plan[:,0])

# the last column of the plan corresponds to the index number in the reference spectra at the given energy 
ind = np.array((plan[:,2]),dtype='int')

# load the reference spectra
ref_spectra = np.loadtxt('fe0_fe3_ref.txt')

# construct the codebook that is used in spectro-tomography computation
codebook = np.hstack((ref_spectra[ind,1].reshape((-1,1)),ref_spectra[ind,2].reshape((-1,1))))

sz = np.shape(proj)
plt.imshow(proj[:,sz[1]//2,:])
plt.title('sinogram')
#
```

```python
# reconstruct the central slice to determine optimal values for rotation axis offset, mu1 and mu2

rot_offset = -0.5
mu1 = 0.05
mu2 = 10
itr_num = 100

sino = proj[:,sz[1]//2,:]
row,col = np.shape(sino)
num_proj, num_state = np.shape(codebook)

rec,err = multistate_tomo_joint_TV(sino,angle,rot_offset,[col,col],codebook,mu1,mu2,itr_num)
for i in range(num_state):
    rec[i][rec[i]<0] = 0


plt.figure(figsize=[15,5])
plt.subplot(131)
plt.imshow(rec[0], interpolation = 'none')
plt.subplot(132)
plt.imshow(rec[1], interpolation = 'none')
plt.subplot(133)
plt.plot(np.log10(err))
```

```python
# reconstruct the entire volumn

num_proj, row, col = np.shape(proj)
num_proj, num_state = np.shape(codebook)

vol_rec = []
for j in range(num_state):
    vol_rec.append(np.zeros((row,col,col)))
for i in tqdm(range(row)):
    sino = proj[:,i,:]
    rec,err = multistate_tomo_joint_TV(sino,angle,rot_offset,[col,col],codebook,mu1,mu2,itr_num)
    for j in range(num_state):
        rec[j][rec[j]<0] = 0
        vol_rec[j][i,:,:] = rec[j]

```

```python
# If reference spectra are unknown, we decompose XANES data into k base vectors
from sklearn.decomposition import TruncatedSVD, PCA, NMF

# load 2D nano-XANES 
nano_XANES_data_path = 'C:\\Users\\hanfe\\OneDrive - Brookhaven National Laboratory\\Experiment\\Yan_2023Q2\\nano-XANES\\'
Fe = tf.imread(f"{nano_XANES_data_path}maps_XRF_Fe_K_cropped_aligned.tif")
Cr = tf.imread(f"{nano_XANES_data_path}maps_XRF_Cr_K_cropped_aligned.tif")

# In 2D nano-XANES measurement, the 52th energy point was not measured
# Since it is far-away from the absorption edge, we duplicate the 51th energy point 
# This was needed because 3D spectr-tomography used 52 energy points
# This only introduce a negligible error

fake_img = Fe[-1,:,:] 
Fe = np.concatenate([Fe,fake_img[np.newaxis,...]],0)
fake_img = Cr[-1,:,:] 
Cr = np.concatenate([Cr,fake_img[np.newaxis,...]],0)

#norm_factor = norm_mass(Cr)
Cr_norm, ratio = normalize_sinogram(Cr)

# normalize Fe to Cr signal, whcih remains a constant over the energy range
sz = np.shape(Cr)
#Cr_norm = Cr*np.tile(np.swapaxes(ratio,0,1)[:,:,np.newaxis],(1,1,sz[2]))

#exclude boarder regions due to misalignment-these areas were not scanned at every energy point  
Fe_norm = Fe[:,2:48,1:41]*np.tile(np.swapaxes(ratio,0,1)[:,:,np.newaxis],(1,1,sz[2]))

sz = Fe_norm.shape
X = np.reshape(np.swapaxes(Fe_norm,0,-1),[-1,sz[0]])

k = 2 # number of components

######different decomposition methods#####

# model = NMF(n_components=k, init='nndsvda', max_iter=2000) 
model = PCA(n_components=k) 

W = model.fit_transform(X) 
H = model.components_

print("W shape:", W.shape) 
print("H shape:", H.shape)

# ---------------------------------------------------- # Plot the learned basis components # ---------------------------------------------------- 
plt.figure(figsize=(10, 5)) 

for i in range(k):
    plt.plot(H[i], label=f"Component {i+1}") 
plt.title(f"Learned basis components using {model.__class__.__name__}") 
plt.legend() 
plt.show()

```

```python
# construct the codebook that was used in spectro-tomography computation
codebook = H.T[ind][:, :k]

rot_offset = -0.5
mu1 = 0.02
mu2 = 1
itr_num = 100
sz = np.shape(proj)
sino = proj[:,sz[1]//2,:]
row,col = np.shape(sino)
num_proj, num_state = np.shape(codebook)

rec,err = multistate_tomo_joint_TV(sino,angle,rot_offset,[col,col],codebook,mu1,mu2,itr_num)
# for i in range(num_state):
#     rec[i][rec[i]<0] = 0


plt.figure(figsize=[15,5])
plt.subplot(131)
for i in range(k):
    plt.plot(H[i], label=f"Component {i+1}") 
plt.title(f"Learned basis components using {model.__class__.__name__}") 
plt.legend() 

plt.subplot(132)
plt.imshow(rec[0])
plt.subplot(133)
plt.imshow(rec[1])
```

```python
# reconstruct the spectrum at each pixel byand fit it with two reference spectra
from scipy.optimize import nnls

recon = np.array(rec).copy()
sz_im = np.shape(recon)
recon = np.reshape(np.swapaxes(recon,0,-1),[-1,sz_im[0]])
recon_spectra = np.reshape(recon @ H, [sz_im[1],sz_im[2],-1])

plt.figure(figsize=[15,5])
plt.subplot(131)
plt.plot(recon_spectra[sz_im[1]//2,sz_im[2]//2,:])
plt.title('Synthesized spectrum at one pixel')


num_energy, num_state = ref_spectra[:,1:3].shape
B = np.reshape(recon_spectra,[-1,num_energy])

# nnls fit to the synthesized spectrum data using two reference spectra
ref = ref_spectra[:,1:3]
num_energy, num_state = np.shape(ref)
c = [nnls(ref,b)[0] for b in B]
state_im = np.reshape(np.swapaxes(c,0,-1),[num_state,sz_im[1],sz_im[2]])

plt.subplot(132)
plt.imshow(state_im[0].transpose())
plt.subplot(133)
plt.imshow(state_im[1].transpose())
```
